```python
contours1, hierarchies2 = cv.findContours(image_canny, cv.RETR_LIST, cv.CHAIN_APPROX_NONE)
contours2, hierarchies2 = cv.findContours(image_lap, cv.RETR_LIST, cv.CHAIN_APPROX_NONE)
contours3, hierarchies2 = cv.findContours(image_sobel_combined, cv.RETR_LIST, cv.CHAIN_APPROX_NONE)
print(len(contours1))
print(len(contours2))
print(len(contours3))
```
